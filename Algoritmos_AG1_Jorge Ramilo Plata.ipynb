### 1. Torres de Hanoi con Divide y vencerás ###

```python
def torres_hanoi(numero_discos, desde, hasta):
    if numero_discos <= 0:
        return 0

    if numero_discos == 1:
        print("Llevar la ficha desde", desde, "hasta", hasta)
        return 1

    auxiliar = 6 - desde - hasta
    movimientos = torres_hanoi(numero_discos - 1, desde, auxiliar)
    print("Llevar la ficha desde", desde, "hasta", hasta)
    movimientos += 1
    movimientos += torres_hanoi(numero_discos - 1, auxiliar, hasta)
    return movimientos


print("Ejemplo con 3 discos:")
total_movimientos = torres_hanoi(3, 1, 3)
print("Número total de movimientos:", total_movimientos)

assert total_movimientos == 7
assert torres_hanoi(0, 1, 3) == 0
print("Pruebas de Hanoi superadas.")
```

### 2. Devolución de cambio por técnica voraz ### 

```python
def cambio_monedas(cantidad, sistema):
    solucion = [0] * len(sistema)
    valor_acumulado = 0

    for i, moneda in enumerate(sistema):
        monedas = (cantidad - valor_acumulado) // moneda
        solucion[i] = monedas
        valor_acumulado += monedas * moneda

        if valor_acumulado == cantidad:
            return solucion

    return None

sistema_habitual = [25, 10, 5, 1]
print("Cambio de 27 con", sistema_habitual, ":", cambio_monedas(27, sistema_habitual))
print("Cambio de 0:", cambio_monedas(0, sistema_habitual))
print("Cantidad 3 con [4, 2]:", cambio_monedas(3, [4, 2]))

assert cambio_monedas(27, sistema_habitual) == [1, 0, 0, 2]
assert cambio_monedas(0, sistema_habitual) == [0, 0, 0, 0]
assert cambio_monedas(3, [4, 2]) is None
print("Pruebas del cambio voraz superadas.")
```

En este ejercicio la función recorre las monedas de mayor a menor, para cada valor lo que hace es calcular cuantas monedas caben sin superar la cantidad, si el sistema no es capaz de formar el importe exacto nos devuelve el valor "None".

### 3. N-Reinas por técnica de vueta atrás ###

```python
def es_prometedora(solucion, etapa):
    for i in range(etapa + 1):
        for j in range(i + 1, etapa + 1):
            misma_fila = solucion[i] == solucion[j]
            misma_diagonal = abs(i - j) == abs(solucion[i] - solucion[j])

            if misma_fila or misma_diagonal:
                return False

    return True


def reinas(numero_reinas):
    soluciones = []
    solucion = [0] * numero_reinas

    def colocar_reina(etapa):
        for fila in range(1, numero_reinas + 1):
            solucion[etapa] = fila

            if es_prometedora(solucion, etapa):
                if etapa == numero_reinas - 1:
                    soluciones.append(solucion.copy())
                else:
                    colocar_reina(etapa + 1)

            solucion[etapa] = 0

    if numero_reinas > 0:
        colocar_reina(0)

    return soluciones


def mostrar_tablero(solucion):
    for fila in range(1, len(solucion) + 1):
        linea = []
        for columna in range(len(solucion)):
            if solucion[columna] == fila:
                linea.append("X")
            else:
                linea.append("-")
        print(" ".join(linea))


soluciones_4 = reinas(4)
soluciones_8 = reinas(8)

print("Soluciones para 4 reinas:", len(soluciones_4))
print(soluciones_4)
print("\nPrimer tablero de 4 reinas:")
mostrar_tablero(soluciones_4[0])
print("\nNúmero de soluciones para 8 reinas:", len(soluciones_8))

assert len(reinas(1)) == 1
assert len(reinas(2)) == 0
assert len(soluciones_4) == 2
assert len(soluciones_8) == 92
print("Pruebas de N-Reinas superadas.")
```

$$ ************ NOTA ************ $$

Como podemos ver, cada posición de la lista representa una columna y su valor indica la final de la reina, se comprueba ante de nada que no haya dos filas iguales ni dos reinas en la misma diagonal, es resulta que una de las posiciones no es válida el algoritmo vuelvo hacia atras y prueba otra fila.

Lo que he realizado en esta mejora de algoritmo ha sido eliminar el argumento mutable de la función inicial, guardar las soluciones y que se muestren sólo los resultados representativos.

### 4. Viaje por el rio. Programación dinámica ###

```python
TARIFAS = [
    [0, 5, 4, 3, 999, 999, 999],
    [999, 0, 999, 2, 3, 999, 11],
    [999, 999, 0, 1, 999, 4, 10],
    [999, 999, 999, 0, 5, 6, 9],
    [999, 999, 999, 999, 0, 999, 4],
    [999, 999, 999, 999, 999, 0, 3],
    [999, 999, 999, 999, 999, 999, 0],
]


def calcular_precios(tarifas):
    numero_embarcaderos = len(tarifas)
    infinito = 9999
    precios = [[infinito] * numero_embarcaderos for _ in range(numero_embarcaderos)]
    ruta = [[-1] * numero_embarcaderos for _ in range(numero_embarcaderos)]

    for i in range(numero_embarcaderos):
        precios[i][i] = 0
        ruta[i][i] = i

    for i in range(numero_embarcaderos - 1):
        for j in range(i + 1, numero_embarcaderos):
            mejor_precio = tarifas[i][j]

            if mejor_precio < 999:
                ruta[i][j] = i
            else:
                mejor_precio = infinito

            for k in range(i + 1, j):
                if tarifas[k][j] < 999:
                    precio_candidato = precios[i][k] + tarifas[k][j]

                    if precio_candidato < mejor_precio:
                        mejor_precio = precio_candidato
                        ruta[i][j] = k

            precios[i][j] = mejor_precio

    return precios, ruta


def reconstruir_ruta(ruta, desde, hasta):
    if desde == hasta:
        return [desde]

    anterior = ruta[desde][hasta]
    if anterior == -1:
        return []

    if anterior == desde:
        return [desde, hasta]

    return reconstruir_ruta(ruta, desde, anterior) + [hasta]


PRECIOS, RUTA = calcular_precios(TARIFAS)
ruta_0_6 = reconstruir_ruta(RUTA, 0, 6)

print("Tabla de precios mínimos:")
for fila in PRECIOS:
    print(fila)

print("\nTabla de rutas:")
for fila in RUTA:
    print(fila)

print("\nPrecio mínimo de 0 a 6:", PRECIOS[0][6])
print("Ruta completa de 0 a 6:", ruta_0_6)

assert PRECIOS[0][6] == 11
assert ruta_0_6 == [0, 2, 5, 6]
assert reconstruir_ruta(RUTA, 6, 0) == []
print("Pruebas del viaje por el río superadas.")
```

$$ ************ NOTA ************ $$

Lo que hace este código es que para cada pareja de embarcaderos se compara el viaje directo con las rutas que mediante un embarcadero intermedio y se reutilizan los mejores precios ya calculados.

La mejora que he llevado a cabo ha sido corregir el código original, ya que este aunque calculaba bien el precio, la función devolvía el valor 0,2,5 y omitía el destino 6, y con esta nueva función reconstruimos la ruta completa y nos dice cuando no existe un trayecto.
